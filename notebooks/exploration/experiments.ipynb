```python
import pandas as pd
import json
from parcsann.utils.dir import get_project_root
from pathlib import Path
```

```python
ROOT_PATH = get_project_root()
PARENT_DIR = ROOT_PATH / "hyperparameters"
EXPERIMENT_DIR = PARENT_DIR / "experiment_2"

directories = [item for item in EXPERIMENT_DIR.iterdir() if item.is_dir()]
TIMESTAMP_ID = 0

params_file = ROOT_PATH / directories[TIMESTAMP_ID] / "all_parameters.json"

with open(params_file, 'r') as f:
    data = json.load(f)

if isinstance(data, dict):
    df = pd.DataFrame([data])
elif isinstance(data, list):
    df = pd.DataFrame(data)

df = pd.concat([df.drop(['params'], axis=1), df['params'].apply(pd.Series)], axis=1)

max_target_row = df.loc[df['target'].idxmax()]
print(max_target_row)
```
